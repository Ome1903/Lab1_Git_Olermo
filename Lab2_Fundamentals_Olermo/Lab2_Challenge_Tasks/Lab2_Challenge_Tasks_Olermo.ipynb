# Exercise 1 - Sensor Monitoring System


```python
ENGINEER = "Olermo"
STUDENT_ID = "TUPM-26-1925"
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")


```

```python
import random

random.seed(SEED_NUM)

sensor_readings = [random.uniform(0, 100) for _ in range(10)]

print("Generated Sensor Data:")
print(sensor_readings)
```

```python
valid_readings = []
invalid_readings = []

for reading in sensor_readings:
    if isinstance(reading, (int, float)):
        valid_readings.append(reading)
    else:
        invalid_readings.append(reading)

print("Valid Readings:", valid_readings)
print("Invalid Readings:", invalid_readings)
```

```python
LOW_LIMIT = 30
HIGH_LIMIT = 70

def classify_reading(value):
    if value <= LOW_LIMIT:
        return "Low"
    elif value <= HIGH_LIMIT:
        return "Normal"
    else:
        return "High"

    for reading in valid_readings:
        print(reading, "->", classify_reading(reading))
```

```python
print("=== EXERCISE 1 SUMMARY ===")
print("Generated Sensor Data:", sensor_readings)
print("Valid Readings:", len(valid_readings))
print("Invalid Readings:", len(invalid_readings))

print("\nClassification Results:")
for reading in valid_readings:
    print(reading, "->", classify_reading(reading))

print("\nExecution Log:")
print("Sensor data generated.")
print("Readings validated.")
print("Valid readings classified.")

print("\nFinal Output:")
print("Sensor monitoring completed.")
```

```python
signal = f"{ENGINEER}{SEED_NUM}" 

print("Generated Signal:", signal)
```

```python
processed_signal = signal.strip().upper()
print("Processed Signal:", processed_signal)
```

```python
letters = 0
digits = 0
spaces = 0
others = 0

for char in processed_signal:
    if char.isalpha():
        letters += 1
        category = "Letter"
    elif char.isdigit():
        digits += 1
        category = "Digit"
    elif char.isspace():
        spaces += 1
        category = "Other"
    print(f"{char} -> {category}")
    
```

```python
print("=== EXERCISE 2 SUMMARY ===")
print("Generated Signal", signal)
print("Processed Signal:", processed_signal)
print("Letters:", letters)
print("Digits:", digits)
print("Spaces:", spaces)
print("Other Characters:", others)

print("\nExecution Log:")
print("Signal generated.")
print("Signal normalized.")
print("Characters analyzed.")

print("\nFinal Output:")
print("Signal analysis completed.")
```

```python
generated_password = ENGINEER + str(SEED_NUM)
print("Generated Password:", generated_password)
```

```python
attempt_limit = 3
attempts_made = 0
authenticated = False
```

```python
while attempts_made < attempt_limit:
    password = input ("Enter password: ")
    attempts_made += 1

    if password == generated_password:
        authenticated = True
        print("Access Granted!")
        break
    else:
        print("Incorrect Password.")
        
```

```python
if authenticated:
    access_result = "Successful"
    final_state = "SYSTEM UNLOCKED"
else:
    access_result = "Failed"
    final_state = "SYSTEM LOCKED"
    
```

```python
print("=== EXERCISE 3 SUMMARY ===")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)
print("Access Result:", access_result)
print("Final System State", final_state)

print("\nExecution Log:")
print("Authentication process started.")
print(f"{attempts_made} attempt(s) processed.")
print("Authentication process completed.")

print("\nFinal Output:")
print(final_state)
```
